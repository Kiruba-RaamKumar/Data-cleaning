#Handle the Null Values

```python
df['winner']=df["winner"].fillna("no winner")
df['winner_runs']=df["winner_runs"].fillna(0)
df['winner_wickets']=df["winner_wickets"].fillna(0)
df["player_of_match"]=df["player_of_match"].fillna("no player")
```

```python
df.isnull().sum()
```

#remove extra space

```python
df["team1"] = df["team1"].str.strip()
df["team2"] = df["team2"].str.strip()
df["venue"] = df["venue"].str.strip()
df["city"] = df["city"].str.strip()
```

#date column

```python
df["winner_runs"] = df["winner_runs"].astype(int)
df["winner_wickets"] = df["winner_wickets"].astype(int)
```
