# To remove extra space

```python
text_columns = df.select_dtypes(include=['object']).columns
df[text_columns] = df[text_columns].apply(lambda x: x.str.strip())
```

# Y its not work column contain a mixture of strings, numbers, or other values

```python
text_columns = df.select_dtypes(include=['object']).columns
df[text_columns] = df[text_columns].apply(
    lambda x: x.map(lambda value: value.strip() if isinstance(value, str) else value))
```
